```python
def search_to_subcat(output_name, catalog, list_of_variables):
    empty_cat = catalog.search()
    empty_cat.serialize(directory=directory, name=output_name, catalog_type="file")
    dfs = []
    for v in list_of_variables:
        dfs.append(catalog.search(realm=v[1], variable_id=v[0]).df)
    concat_df = pd.concat(dfs, ignore_index=True)
    concat_df.to_csv(directory+output_name+".csv", header=False, index=None, mode='a')
    return
```

```python
def search_to_df(catalog, list_of_variables):
    dfs = []
    for v in list_of_variables:
        dfs.append(catalog.search(realm=v[1], variable_id=v[0]).df)
    concat_df = pd.concat(dfs, ignore_index=True)
    return concat_df
```
